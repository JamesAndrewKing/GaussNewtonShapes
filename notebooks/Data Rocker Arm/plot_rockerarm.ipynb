```python
import torch
import numpy as np
import matplotlib.pyplot as plt

# Load the loss data
loss_over_iters_adam = torch.load("loss_over_time_rockerarm_adam.pth")
loss_over_iters_gn = torch.load("loss_over_time_rockerarm_gn.pth")
loss_over_iters_lbfgs = torch.load("loss_over_time_rockerarm_lbfgs.pth")

# Extract keys (iterations) and values (losses)
adam_iterations = np.array(list(loss_over_iters_adam.keys()))
adam_values = np.array(list(loss_over_iters_adam.values()))

gn_iterations = np.array(list(loss_over_iters_gn.keys()))
gn_values = np.array(list(loss_over_iters_gn.values()))

lbfgs_iterations = np.array(list(loss_over_iters_lbfgs.keys()))
lbfgs_values = np.array(list(loss_over_iters_lbfgs.values()))

fig, ax = plt.subplots(1, 1, figsize=(5, 3))

cmap = plt.get_cmap('tab10')
ax.plot(adam_iterations, adam_values, label="Adam", c=cmap(0))
ax.plot(lbfgs_iterations, lbfgs_values, label="LBFGS", c=cmap(1))
ax.plot(gn_iterations, gn_values, label="GN", c=cmap(2))

ax.set_xlabel("Time in [s]")
ax.set_ylabel("Loss")
ax.semilogy()
ax.legend()
plt.tight_layout()
plt.savefig("loss_over_time_rockerarm.png", dpi=300)
plt.show()
```
